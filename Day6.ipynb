```python
# import nec libs
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score
```

```python
# load 
wdf = pd.read_csv(r"C:\Users\sahan\Downloads\framingham.csv\Lab 22 winequality-red.csv")
wdf.head()
```

```python
def log_transform(col):
    return np.log(col)
wdf['residual sugar']=wdf[['residual sugar']].apply(log_transform, axis=1)
wdf['chlorides']=wdf[['chlorides']].apply(log_transform, axis=1)
wdf['free sulphur dioxide']=wdf[['free sulphur dioxide']].apply(log_transform, axis=1)
wdf['total sulphur dioxide']=wdf[['total sulphur dioxide']].apply(log_transform, axis=1)
wdf['sulphates']=wdf[['sulphates']].apply(log_transform, axis=1)
```

```python
# split the dataset - dv y - idv x
x = wdf.drop(columns=['quality'])
y = wdf['quality']
```

```python
from imblearn.over_sampling import SMOTE
oversample = SMOTE(k_neighbors=4)
x, y = oversample.fit_resample(x.fillna(0),y)
```

```python
y.value_counts()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,
                                               test_size=0.2,
                                               random_state=42)
```

```python
x_train.shape, x_test.shape
```

```python
# 1.logistic regression - sigmoid function
lr_model = LogisticRegression()
lr_model.fit(x_train, y_train)
lr_pred = lr_model.predict(x_test)
lr_pred
```

```python
# accuraccy
acc = accuracy_score(y_test, lr_pred)
acc*100
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,lr_pred)
cm
```

```python
rfc_model = RandomForestClassifier()
rfc_model.fit(x_train, y_train)
rfc_pred = rfc_model.predict(x_test)
acc_rfc = accuracy_score(y_test, rfc_pred)
acc_rfc*100
```

# Unsupervised ML - K Means Clutering

```python
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv(r"C:\Users\sahan\Downloads\framingham.csv\Lab 23 Wholesale customers data.csv")
df.head()
```

```python
df.describe()
```

```python
# standerdization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
pd.DataFrame(df_scaled)
```

```python
# define kmeans and train it 
kmeans = KMeans(n_clusters=2, init='k-means++')
# training
kmeans.fit(df_scaled)
```

```python
SSE = []
for cluster in range(1,20):
    kmeans = KMeans(n_clusters=cluster, init='k-means++')
    kmeans.fit(df_scaled)
    SSE.append(kmeans.inertia_)

frame = pd.DataFrame({'Cluster':range(1,20),'SSE':SSE})
plt.figure(figsize=(12,6))
plt.plot(frame['Cluster'],frame['SSE'],marker='o')
plt.xlabel('Number of Cluster')
plt.ylabel('Inertia')
plt.show()
```

```python
# for k = 5
kmeans = KMeans(n_clusters=5, init='k-means++')
kmeans.fit(df_scaled)
pred = kmeans.predict(df_scaled)
```

```python
frame = pd.DataFrame(df_scaled)
frame['cluster']= pred
frame['cluster'].value_counts()
```

```python
kmeans = KMeans(n_clusters=7, init='k-means++')
kmeans.fit(df_scaled)
pred = kmeans.predict(df_scaled)
frame = pd.DataFrame(df_scaled)
frame['cluster']= pred
frame['cluster'].value_counts()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, InputLayer, Dropout
```

```python
model1 = Sequential()
model1.add(InputLayer(input_shape=(30,))) 
model1.add(Dense(26,activation='relu'))
model1.add(Dense(15,activation='relu'))
model1.add(Dense(1,activation='sigmoid'))
model1.summary()
```

```python
model1.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

```python
model1.fit(x=x_tr, y=y_tr, epochs=100, validation_data=(x_te,y_te))
```

```python
model_loss = pd.DataFrame(model1.history.history)
model_loss.plot
```
